```python
# we will join title overview, genres to convert them into vectorization 
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['title'])
```

```python
df['overview'] = df['overview'].fillna(' ')
```

```python
df.iloc[0]['genres']
```

```python
import ast
# ast.literal_eval is used to get the name property from the dictionary as well as the nested dictionary
df['genres'] = df['genres'].apply(lambda x: " ".join([i['name'] for i in ast.literal_eval(x)]))
```

```python
df.head()
```

```python
# filling tagline 

df['tagline'] = df['tagline'].fillna(' ')
```

```python
df.isnull().sum()
```

```python
# now combining three columns 

df['tags'] = df['overview']+" "+df['genres']+" "+df['tagline']
```

```python
df.head()
```

```python
# converting tags into numbers /tokens 

df['tags'][0]
```

```python
# 1. removing stopwords 

import nltk 
from nltk.corpus import stopwords 
```

```python
from nltk.stem import WordNetLemmatizer 
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_txt(txt):
    txt= str(txt).lower() # converting txt to lowercase
    txt = re.sub(r'[^a-zA-Z\s]','',txt) # removing punctuations 
    # tokenizing the words 
    words = txt.split() 
    # lemmatization 
    words = [word for word in words if word not in stop_words]
    
    #  lemmatize 
    words = [lemmatizer.lemmatize(word) for word in words] 

    return " ".join(words) 
```

```python
df['tags'] = df['tags'].apply(preprocess_txt)
```

```python

df['tags'][0]
```

```python
# 2. convert these tokens into vectors using tf-idf 

# a. reseting the index 
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index = df['title']).drop_duplicates()
```

```python
indices
```

```python
# applying vectorization 

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf_vectorizer = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf_vectorizer.fit_transform(df['tags']) # matrix size would be [45447 X 50000]
```

```python
tfidf_matrix
```

```python
# we will use cosine similarity to predict the similar movie/vector 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title,n=10): # here n is the no of recommendations that would be provided based on titlte 
    if title not in indices:
        return ['Movie not found']
    idx = indices[title] 
    # cosine similarity 
    # 1. get the similarity score 
    sim_score = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten() 
    similar_idx = sim_score.argsort()[::-1][1:n+1] 
    return df['title'].iloc[similar_idx]

    
```

```python
recommend('Toy Story')
```

```python
recommend('Jumanji')
```

```python
recommend('Avatar')
```
